# **2. Add 4 hidden layers with the ReLU activation and 16 neurons for the fourth layer.**

```python
# Add 4 hidden layers
model = tf.keras.Sequential([tf.keras.layers.Dense(64, activation='relu', input_shape=(X_train.shape[1],)),
                           tf.keras.layers.Dense(32, activation='relu'),
                           tf.keras.layers.Dense(32, activation='relu'),
                           tf.keras.layers.Dense(32, activation='relu'),
                           tf.keras.layers.Dense(32, activation='relu'),
                           tf.keras.layers.Dense(16, activation='relu'),
                           tf.keras.layers.Dense(1, activation='sigmoid')])
```

# **3. Compile the model with binary_crossentropy as loss, Adam optimiser, and print the accuracy of the model.**

# **6. Create a vector of different batch_sizes=np.array([16, 32, 64]) and loop through it, retraining the model each time, and print the performances. Use the same model and number of epochs.**

```python
# Different bach sizes
batch_sizes = np.array([16, 32, 64])

for i in batch_sizes:
  epochs = 10
  model.fit(X_train, y_train, epochs=epochs, batch_size=i, validation_data=(X_val, y_val))
  score = model.evaluate(X_test, y_test)
  print(f"Batch size: {i}, Accuracy: {score[1]}, Loss: {score[0]}")
```

The bach size 16 accuracy and loss results performed better. As batch size decreases, accuracy increases and loss decreases

# **7. Create a vector of different epochs=np.array([10, 20, 30]) and loop through it, retraining the model each time and using the batch size. Jot down which model gave you the highest accuracy.**

```python
# Different epochs sizes
epochs_array = np.array([10, 20, 30])
best_batch_size = 16

for n in epochs_array:
  epochs = n
  model.fit(X_train, y_train, epochs=epochs, batch_size=best_batch_size, validation_data=(X_val, y_val))
  score = model.evaluate(X_test, y_test, verbose=0)
  print('\nEpochs:', n)
  print('Test loss:', score[0])
  print('Test accuracy:', score[1])
```
