# **1. Continue to work on the model you created in Activity 3.2.3.**

```python
# Import necessary libraries
import numpy as np
import tensorflow as tf
import pandas as pd

from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
```

```python
# Start your activity here:

# URL to import data set from GitHub.
url = '<path-to-dataset>/spamdata.csv'
```

```python
# create a dataframe
df = pd.read_csv(url, header=None)
```

```python
# explore and view the dataset
df.head()
```

```python
# Specify features
X = df.iloc[:, :-1]
y = df.iloc[:, -1]
```

```python
# Split the data into test and train data sets.
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

X_train, X_val, y_train, y_val = train_test_split(X_train, y_train, test_size=0.1)
```

```python
# Standarized the features
scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_val = scaler.transform(X_val)
X_test = scaler.transform(X_test)
```

```python
# Add 4 hidden layers
model = tf.keras.Sequential([tf.keras.layers.Dense(64, activation='relu', input_shape=(X_train.shape[1],)),
                           tf.keras.layers.Dense(32, activation='relu'),
                           tf.keras.layers.Dense(32, activation='relu'),
                           tf.keras.layers.Dense(32, activation='relu'),
                           tf.keras.layers.Dense(32, activation='relu'),
                           tf.keras.layers.Dense(16, activation='relu'),
                           tf.keras.layers.Dense(1, activation='sigmoid')])
```

```python
# Compile the model
model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
```

```python
# Train the model
history=model.fit(X_train, y_train, epochs=10, batch_size=64, validation_data=(X_val, y_val))
```

```python
# Evaluate the model
loss, accuracy = model.evaluate(X_test, y_test)
```

```python
# Different bach sizes
batch_sizes = np.array([16, 32, 64])

for i in batch_sizes:
  epochs = 10
  model.fit(X_train, y_train, epochs=epochs, batch_size=i, validation_data=(X_val, y_val))
  score = model.evaluate(X_test, y_test)
  print(f"Batch size: {i}, Accuracy: {score[1]}, Loss: {score[0]}")
```

```python
# Different epochs sizes
epochs_array = np.array([10, 20, 30])
best_batch_size = 16

for n in epochs_array:
  epochs = n
  model.fit(X_train, y_train, epochs=epochs, batch_size=best_batch_size, validation_data=(X_val, y_val))
  score = model.evaluate(X_test, y_test, verbose=0)
  print('\nEpochs:', n)
  print('Test loss:', score[0])
  print('Test accuracy:', score[1])
```

# **2. Select the best model you obtained through hyperparameter tuning. Substantiate your choice.**

The best model was the one with 4 hidden layers as it increases the accuracy to 95%. Batch size was 16  with 94.1% accuracy followed by 10 epochs where resulted in 94.3%.
The choice for it is based on which model gave us the high accuracy and imporved the user experience by selecting the correct spam email.

# **3. Run the chosen model again and save it in an h5 file named best_model.h5. Remember to specify the path.**

```python
# Best model
best_model = tf.keras.Sequential([tf.keras.layers.Dense(64, activation='relu', input_shape=(X_train.shape[1],)),
                           tf.keras.layers.Dense(32, activation='relu'),
                           tf.keras.layers.Dense(32, activation='relu'),
                           tf.keras.layers.Dense(32, activation='relu'),
                           tf.keras.layers.Dense(32, activation='relu'),
                           tf.keras.layers.Dense(16, activation='relu'),
                           tf.keras.layers.Dense(1, activation='sigmoid')])
```

```python
# Best epochs
best_epochs = 10

# Best batch sizes
best_batch_size = 16
```

```python
# Best model compile
best_model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
```

```python
# Best model fit
best_model.fit(X_train, y_train, epochs=best_epochs, batch_size=best_batch_size, validation_data=(X_val, y_val))
```

```python
# Save best model
best_model.save('best_model.h5')
```

# **4. Check further metrics for the model with the predict function applied to your model variable in order to create predictions on the X_test data set.**

```python
# Check further metrics for the model
y_pred = best_model.predict(X_test)
```

# **5. Convert probabilities to binary predictions and print the accuracy, F1 score, and recall. You can use the following code:**

*  predictions: y_pred = (y_pred > 0.5).astype(int)
*  confusion matrix metrics: accuracy_score, precision_score, recall_score and f1_score functions.

```python
# Convert probability to binary pred
y_pred = (y_pred > 0.5).astype(int)
```

```python
# Import libraries for accuracy, F1 score and recall
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
```

```python
# Results accuracy, F1 score and recall
print('Accuracy:', accuracy_score(y_test, y_pred))
print('Precision:', precision_score(y_test, y_pred))
print('Recall:', recall_score(y_test, y_pred))
print('F1 Score:', f1_score(y_test, y_pred))
```
