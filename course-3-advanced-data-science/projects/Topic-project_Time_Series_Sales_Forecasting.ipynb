## Classical techniques: SARIMA

```python
# Build the two clean weekly series for modelling
def get_series(isbn):
    s = weekly_post2012[weekly_post2012['ISBN'] == isbn].set_index('End Date')['Volume'].sort_index()
    s = s.asfreq('W')
    return s

caterpillar = get_series('9780241003008')
alchemist = get_series('9780722532935')

print("caterpillar:", len(caterpillar), "| alchemist:", len(alchemist))
```

```python
# Tests the multiplicative vs additive for both books
def compare_decomposition(series, name):
    print(f"\n{'='*55}\n{name}\n{'='*55}")

    # Test 1: mean-variance relationship grouped by year
    yearly = series.groupby(series.index.year).agg(['mean', 'std'])
    corr = yearly['mean'].corr(yearly['std'])
    print("Mean-variance test (correlation of yearly mean vs yearly std):")
    print(f"  correlation = {corr:.3f}")
    print("  -> strong positive corr suggests MULTIPLICATIVE; near zero suggests ADDITIVE")

    # Test 2: fit both, compare residual variance
    s_pos = series.replace(0, 0.1)
    add = seasonal_decompose(series, model='additive', period=52)
    mul = seasonal_decompose(s_pos, model='multiplicative', period=52)
    add_resid_var = np.nanvar(add.resid)
    mul_resid_var = np.nanvar(mul.resid)
    print("\nResidual variance (lower = better fit):")
    print(f"  additive resid var       = {add_resid_var:.4f}")
    print(f"  multiplicative resid var = {mul_resid_var:.4f}")

    return yearly, corr

cat_year, cat_corr = compare_decomposition(caterpillar, "The Very Hungry Caterpillar")
alc_year, alc_corr = compare_decomposition(alchemist, "The Alchemist")
```

The mean variance is higher for Caterpillar (mean variance corr = 0.436; leans multiplicative, but only moderately) and Alchemist (mean variance corr = 0.149; leans additive, weakly); therefore, neither is a strong signal.

To choose between additive and multiplicative decomposition objectively rather than by visual assumption, I will decompose both books using both methods and compare the results. The deciding criterion is the residual component.

```python
# Test both decompose methods to understand the sasonality
def compare_both_ways(series, name):
    s_pos = series.replace(0, 0.1)
    add = seasonal_decompose(series, model='additive', period=52)
    mul = seasonal_decompose(s_pos, model='multiplicative', period=52)

    fig, axes = plt.subplots(4, 2, figsize=(16, 12))
    fig.suptitle(f'{name}: Additive (left) vs Multiplicative (right)', y=1.00, fontsize=14)

    for col, (dec, label) in enumerate([(add, 'Additive'), (mul, 'Multiplicative')]):
        axes[0, col].plot(dec.observed); axes[0, col].set_title(f'{label} — Observed')
        axes[1, col].plot(dec.trend);    axes[1, col].set_title('Trend')
        axes[2, col].plot(dec.seasonal); axes[2, col].set_title('Seasonal')
        axes[3, col].plot(dec.resid);    axes[3, col].set_title('Residual')
    plt.tight_layout()
    plt.show()

compare_both_ways(caterpillar, 'The Very Hungry Caterpillar')
compare_both_ways(alchemist, 'The Alchemist')
```

Since the choice of decomposition method was not obvious from the series alone, the analyst applied both methods (multiplicative and additive) to each book to compare the residuals over time.

The Very Hungry Caterpillar:

- Trend: a clear long-run upward trend (range around 700 to 2100 weekly units), with a visible dip during the 2020 - 2021 COVID period followed by a strong recovery.
- Seasonal: a sharp annual peak at Christmas (December).
- Residual: the additive residual widened as the sales level rose, indicating fluctuations scale with the trend (hence multiplicative); the 2020-2021 zero-sales weeks appear as contamination in this period.

For Caterpillar, the multiplicative residuals show a more stable, level-independent spread across time. In contrast, the additive residuals widen as the sales level rises, indicating that seasonal fluctuations scale with the trend, making the multiplicative residuals the better solution.

The Alchemist:

- Trend: a flatter, more cyclical trend (roughly 400 to 700 weekly units) rather than sustained growth, with a pronounced dip in 2020-2021 that coincides with the COVID period.
- Seasonal: a sharp annual peak at Christmas, in December.
- Residual: the additive residual kept stable, spread across the whole period, confirming a seasonal swing.

For The Alchemist, the additive residuals remain stable and spread across the whole period; although there is a December peak, they stay roughly constant throughout, making the multiplicative method the appropriate choice.

```python
# Stationarity check: ADF test
from statsmodels.tsa.stattools import adfuller

def adf_test(series, name):
    result = adfuller(series.dropna())
    print(f"{name}")
    print(f"  ADF statistic: {result[0]:.4f}")
    print(f"  p-value: {result[1]:.4f}")
    print(f"  -> {'STATIONARY' if result[1] < 0.05 else 'NON-STATIONARY'} (at 5% level)")
    print()

# Raw series
adf_test(caterpillar, "Caterpillar (raw)")
adf_test(alchemist, "Alchemist (raw)")

# Differenced to see if achieves stationarity
adf_test(caterpillar.diff(), "Caterpillar (differenced once)")
adf_test(alchemist.diff(), "Alchemist (differenced once)")
```

The ADF test is based on a stochastic trend but does not account for seasonality. As a result, the ADF results have a blind spot indicating that the data is stationary; however, seasonality (and non-stationarity) was already visible in earlier evidence: the raw plots and the slowly decaying ACF.

```python
# Libraries for XGBoost
from sktime.forecasting.compose import make_reduction
from sktime.forecasting.model_selection import ForecastingGridSearchCV
from sktime.split import ExpandingWindowSplitter
from xgboost import XGBRegressor
import numpy as np
print("sktime imports OK")
```

```python
# Cross-validation setup to define forecast horizon of 32 weeks and expanding window splitter
H = 32
fh = np.arange(1, H + 1)

def make_cv(train_len):
    initial = int(train_len * 0.7)
    return ExpandingWindowSplitter(fh=fh, initial_window=initial, step_length=H)
```

```python
# Pipeline and cross-validation, grid search including window_length
def tune_sktime_xgb(train, name):
    print(f"\n{'='*55}\n{name}\n{'='*55}")

    regressor = XGBRegressor(objective='reg:squarederror', random_state=42)
    forecaster = make_reduction(regressor, strategy="recursive")

    param_grid = {
        "window_length": [4, 8, 12, 26],
        "estimator__n_estimators": [100, 300],
        "estimator__max_depth": [3, 5],
        "estimator__learning_rate": [0.05, 0.1],
    }

    cv = make_cv(len(train))
    gscv = ForecastingGridSearchCV(forecaster, cv=cv, param_grid=param_grid, n_jobs=-1)
    gscv.fit(train)

    print("Best params:", gscv.best_params_)
    return gscv
```

```python
# Create an LSTM model for both books.
from sklearn.preprocessing import MinMaxScaler

def prepare_lstm_data(train, window_length=8):
    scaler = MinMaxScaler()
    scaled = scaler.fit_transform(train.values.reshape(-1, 1)).flatten()

    # Build windowed sequences
    X, y = [], []
    for i in range(window_length, len(scaled)):
        X.append(scaled[i - window_length:i])
        y.append(scaled[i])
    X = np.array(X).reshape(-1, window_length, 1)
    y = np.array(y)
    return X, y, scaler
```

```python
# Reproducibility setup: fix random seeds before tuning
import os, random
import numpy as np
import tensorflow as tf

SEED = 42
os.environ['PYTHONHASHSEED'] = str(SEED)
random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

print("Seeds set to", SEED)
```

```python
# Apply KerasTuner, and perform hyperparameter tuning using the training data, for both books.
import keras_tuner as kt
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from tensorflow.keras.optimizers import Adam

def build_tunable_lstm(hp):
    model = Sequential()
    model.add(LSTM(
        units=hp.Int('units', min_value=32, max_value=128, step=32),
        activation='relu',
        input_shape=(8, 1)
    ))
    model.add(Dropout(hp.Float('dropout', min_value=0.1, max_value=0.4, step=0.1)))
    model.add(Dense(1))
    model.compile(
        optimizer=Adam(hp.Choice('lr', values=[0.001, 0.005, 0.01])),
        loss='mse'
    )
    return model
```

```python
def tune_lstm(X, y, name, max_trials=20):
    tuner = kt.RandomSearch(
        build_tunable_lstm,
        objective='val_loss',
        max_trials=max_trials,
        executions_per_trial=1,
        overwrite=True,
        directory='kt_dir',
        project_name=name.replace(' ', '_'),
        seed=SEED
    )
    tuner.search(X, y, epochs=50, validation_split=0.2, verbose=0)
    best_hp = tuner.get_best_hyperparameters(1)[0]
    print(f"{name} best hyperparameters:")
    print(f"  units:   {best_hp.get('units')}")
    print(f"  dropout: {best_hp.get('dropout')}")
    print(f"  lr:      {best_hp.get('lr')}")
    best_model = tuner.get_best_models(1)[0]
    return tuner, best_hp, best_model
```

```python
# Use the best model to predict the final 32 weeks of data for both books. Plot original data and MAE/MAPE
def forecast_lstm(model, train, test, scaler, name, window_length=8, H=32):
    last_window = scaler.transform(train.values[-window_length:].reshape(-1, 1)).flatten()

    preds_scaled = []
    current = last_window.copy()
    for _ in range(H):
        x_input = current.reshape(1, window_length, 1)
        next_scaled = model.predict(x_input, verbose=0)[0, 0]
        preds_scaled.append(next_scaled)
        current = np.append(current[1:], next_scaled)

    # Inverse-scale back to real sales units
    preds = scaler.inverse_transform(np.array(preds_scaled).reshape(-1, 1)).flatten()
    y_pred = pd.Series(preds, index=test.index)

    mae = mean_absolute_error(test, y_pred)
    mape = mean_absolute_percentage_error(test, y_pred)

    plt.figure(figsize=(14, 5))
    plt.plot(train.index[-100:], train.values[-100:], label='Train (recent)', color='steelblue')
    plt.plot(test.index, test.values, label='Actual', color='black', marker='o', markersize=3)
    plt.plot(test.index, y_pred.values, label='LSTM Forecast', color='purple')
    plt.title(f'{name} — LSTM 32-week Forecast (MAE={mae:.1f}, MAPE={mape:.2%})')
    plt.xlabel('Date'); plt.ylabel('Volume'); plt.legend(); plt.grid(True)
    plt.tight_layout(); plt.show()

    print(f"{name}: MAE={mae:.2f}, MAPE={mape:.2%}")
    return y_pred, mae, mape
```

```python
# Prepare SARIMA residuals
import numpy as np
import pandas as pd

def get_sarima_residuals(model, train, name):
    fitted = model.predict_in_sample()
    fitted = pd.Series(fitted, index=train.index)
    residuals = train - fitted
    print(f"{name}: residuals computed, {len(residuals)} points, "
          f"mean={residuals.mean():.2f}, std={residuals.std():.2f}")
    return residuals

cat_resid = get_sarima_residuals(caterpillar_arima, cat_train, "Caterpillar")
alc_resid = get_sarima_residuals(alchemist_arima, alc_train, "Alchemist")
```

```python
# Prepare residuals for LSTM (scale and window)
from sklearn.preprocessing import MinMaxScaler

def prepare_residuals_for_lstm(residuals, window_length=8):
    scaler = MinMaxScaler()
    scaled = scaler.fit_transform(residuals.values.reshape(-1, 1)).flatten()
    X, y = [], []
    for i in range(window_length, len(scaled)):
        X.append(scaled[i - window_length:i])
        y.append(scaled[i])
    X = np.array(X).reshape(-1, window_length, 1)
    y = np.array(y)
    return X, y, scaler

X_cat_res, y_cat_res, cat_res_scaler = prepare_residuals_for_lstm(cat_resid)
X_alc_res, y_alc_res, alc_res_scaler = prepare_residuals_for_lstm(alc_resid)

print("Caterpillar residual X:", X_cat_res.shape)
print("Alchemist residual X:", X_alc_res.shape)
```

```python
# Train the LSTM on the SARIMA residuals
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout

def train_residual_lstm(X, y, name, window_length=8):
    model = Sequential([
        LSTM(64, activation='relu', input_shape=(window_length, 1)),
        Dropout(0.1),
        Dense(1)
    ])
    model.compile(optimizer='adam', loss='mse')
    print(f"Training residual LSTM for {name}...")
    model.fit(X, y, epochs=50, batch_size=16, validation_split=0.2, verbose=0)
    print(f"{name}: residual LSTM trained.")
    return model
```

```python
cat_res_lstm = train_residual_lstm(X_cat_res, y_cat_res, "Caterpillar")
```

```python
alc_res_lstm = train_residual_lstm(X_alc_res, y_alc_res, "Alchemist")
```

```python
# Recursively forecast 32 weeks of residuals with the LSTM
def forecast_residuals(res_lstm, res_scaler, residuals, index, window_length=8, H=32):
    last_window = res_scaler.transform(residuals.values[-window_length:].reshape(-1, 1)).flatten()
    preds_scaled = []
    current = last_window.copy()
    for _ in range(H):
        x_input = current.reshape(1, window_length, 1)
        next_scaled = res_lstm.predict(x_input, verbose=0)[0, 0]
        preds_scaled.append(next_scaled)
        current = np.append(current[1:], next_scaled)

    res_fc = res_scaler.inverse_transform(np.array(preds_scaled).reshape(-1, 1)).flatten()
    return pd.Series(res_fc, index=index)
```

```python
cat_res_fc = forecast_residuals(cat_res_lstm, cat_res_scaler, cat_resid, cat_test.index)
alc_res_fc = forecast_residuals(alc_res_lstm, alc_res_scaler, alc_resid, alc_test.index)
print("Residual forecasts ready")
```

```python
# Evaluate (MAE/MAPE) and plot
def evaluate_hybrid(sarima_fc, hybrid_fc, train, test, name):
    mae = mean_absolute_error(test, hybrid_fc)
    mape = mean_absolute_percentage_error(test, hybrid_fc)
    plt.figure(figsize=(14, 5))
    plt.plot(train.index[-100:], train.values[-100:], label='Train (recent)', color='steelblue')
    plt.plot(test.index, test.values, label='Actual', color='black', marker='o', markersize=3)
    plt.plot(test.index, sarima_fc.values, label='SARIMA only', color='red', alpha=0.5, linestyle='--')
    plt.plot(test.index, hybrid_fc.values, label='Sequential Hybrid', color='darkorange')
    plt.title(f'{name} — Sequential Hybrid (MAE={mae:.1f}, MAPE={mape:.2%})')
    plt.xlabel('Date'); plt.ylabel('Volume'); plt.legend(); plt.grid(True)
    plt.tight_layout(); plt.show()
    print(f"{name}: Hybrid MAE={mae:.2f}, MAPE={mape:.2%}")
    return mae, mape
```

```python
# Parallel Hybrid: weighted average of SARIMA and LSTM forecasts
def parallel_hybrid(sarima_fc, lstm_fc, test, name, w=0.5):
    hybrid_fc = w * sarima_fc + (1 - w) * lstm_fc
    mae = mean_absolute_error(test, hybrid_fc)
    mape = mean_absolute_percentage_error(test, hybrid_fc)
    print(f"{name} (w={w}): MAE={mae:.2f}, MAPE={mape:.2%}")
    return hybrid_fc, mae, mape
```

```python
# Parallel Hybrid: sweep weights to find the best combination
def tune_weight(sarima_fc, lstm_fc, test, name):
    weights = np.arange(0, 1.01, 0.1)
    results = []
    for w in weights:
        hybrid = w * sarima_fc + (1 - w) * lstm_fc
        mae = mean_absolute_error(test, hybrid)
        results.append((round(w, 1), round(mae, 2)))

    print(f"\n{name} — MAE by weight (w = SARIMA share):")
    for w, mae in results:
        print(f"  w={w}: MAE={mae}")

    best_w, best_mae = min(results, key=lambda r: r[1])
    print(f"  BEST: w={best_w} (MAE={best_mae})")
    return results, best_w
```

```python
cat_weights, cat_best_w = tune_weight(cat_sarima_fc, cat_lstm_pred, cat_test, "Caterpillar")
```

```python
alc_weights, alc_best_w = tune_weight(alc_sarima_fc, alc_lstm_pred, alc_test, "Alchemist")
```

```python
# Parallel Hybrid: plot best-weight forecast vs actual
def plot_parallel_hybrid(sarima_fc, lstm_fc, test, train, best_w, name):
    hybrid_fc = best_w * sarima_fc + (1 - best_w) * lstm_fc
    mae = mean_absolute_error(test, hybrid_fc)
    mape = mean_absolute_percentage_error(test, hybrid_fc)

    plt.figure(figsize=(14, 5))
    plt.plot(train.index[-100:], train.values[-100:], label='Train (recent)', color='steelblue')
    plt.plot(test.index, test.values, label='Actual', color='black', marker='o', markersize=3)
    plt.plot(test.index, hybrid_fc.values, label=f'Parallel Hybrid (w={best_w})', color='teal')
    plt.title(f'{name} — Parallel Hybrid (MAE={mae:.1f}, MAPE={mape:.2%})')
    plt.xlabel('Date'); plt.ylabel('Volume'); plt.legend(); plt.grid(True)
    plt.tight_layout(); plt.show()
```

```python
# Parallel Hybrid: plot MAE across weights (shows why the best weight was chosen)
def plot_weight_sweep(results, name):
    ws = [r[0] for r in results]
    maes = [r[1] for r in results]
    plt.figure(figsize=(10, 4))
    plt.plot(ws, maes, marker='o', color='purple')
    best_w = ws[np.argmin(maes)]
    plt.axvline(best_w, color='red', linestyle='--', label=f'Best w={best_w}')
    plt.title(f'{name} — MAE vs SARIMA weight')
    plt.xlabel('Weight on SARIMA (w)'); plt.ylabel('MAE'); plt.legend(); plt.grid(True)
    plt.tight_layout(); plt.show()
```

```python
# Aggregate the weekly sales data to monthly sales data for both books.
cat_monthly = caterpillar.resample('ME').sum()
alc_monthly = alchemist.resample('ME').sum()

print("Caterpillar monthly:", len(cat_monthly), "months")
print(cat_monthly.head())
print("\nAlchemist monthly:", len(alc_monthly), "months")
print(alc_monthly.tail())
```

```python
# Train the XGBoost model on this data.
from sktime.forecasting.compose import make_reduction
from sktime.forecasting.model_selection import ForecastingGridSearchCV
from sktime.split import ExpandingWindowSplitter

# 8 month forecast horizon
H_m = 8
fh_m = np.arange(1, H_m + 1)

def tune_monthly_xgb(train, name):
    print(f"\n{'='*50}\n{name} (monthly XGBoost)\n{'='*50}")
    regressor = XGBRegressor(objective='reg:squarederror', random_state=42)
    forecaster = make_reduction(regressor, strategy="recursive")

    param_grid = {
        "window_length": [3, 6, 12],
        "estimator__n_estimators": [100, 300],
        "estimator__max_depth": [3, 5],
        "estimator__learning_rate": [0.05, 0.1],
    }
    cv = ExpandingWindowSplitter(fh=fh_m, initial_window=int(len(train)*0.7), step_length=H_m)
    gscv = ForecastingGridSearchCV(forecaster, cv=cv, param_grid=param_grid, n_jobs=-1)
    gscv.fit(train)
    print("Best params:", gscv.best_params_)
    return gscv
```

```python
# Sanity check The forecast horizon is 8 months.
H_m = 8
cat_m_train, cat_m_test = cat_monthly[:-H_m], cat_monthly[-H_m:]
alc_m_train, alc_m_test = alc_monthly[:-H_m], alc_monthly[-H_m:]
print("Monthly train/test:", len(cat_m_train), len(cat_m_test))
```

```python
cat_m_xgb = tune_monthly_xgb(cat_m_train, "Caterpillar")
```

```python
alc_m_xgb = tune_monthly_xgb(alc_m_train, "Alchemist")
```

```python
# Forecast using the trained model.
def forecast_monthly(gscv, train, test, name, model_label):
    y_pred = gscv.predict(fh=fh_m)
    y_pred.index = test.index

    mae = mean_absolute_error(test, y_pred)
    mape = mean_absolute_percentage_error(test, y_pred)

    plt.figure(figsize=(12, 5))
    plt.plot(train.index[-24:], train.values[-24:], label='Train (recent)', color='steelblue')
    plt.plot(test.index, test.values, label='Actual', color='black', marker='o')
    plt.plot(test.index, y_pred.values, label=f'{model_label} Forecast', color='green', marker='s')
    plt.title(f'{name} — Monthly {model_label} (MAE={mae:.1f}, MAPE={mape:.2%})')
    plt.xlabel('Date'); plt.ylabel('Volume'); plt.legend(); plt.grid(True)
    plt.tight_layout(); plt.show()

    print(f"{name} monthly {model_label}: MAE={mae:.2f}, MAPE={mape:.2%}")
    return y_pred, mae, mape
```
