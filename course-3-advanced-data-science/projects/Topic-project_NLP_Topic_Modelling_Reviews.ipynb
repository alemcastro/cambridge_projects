# Word Clouds & Word Frequency

```python
# Perform preprocessing of the data - change to lower case, remove stopwords using NLTK, and remove numbers.
nltk.download('stopwords')
nltk.download('punkt')
nltk.download('punkt_tab')
stop_words = set(stopwords.words('english'))

def preprocess_text(text):
    text = text.lower()
    text = re.sub(r'\d+', '', text)
    tokens = word_tokenize(text)
    tokens = [word for word in tokens if word not in stop_words and word not in string.punctuation]
    return tokens
```

# BERTopic - Modelling

```python
# With the data frame created in the previous step:
# Filter out the reviews that are from the locations common to both data sets.
df_google_negative_common = df_google_negative[df_google_negative['Location Name'].isin(common_locations)]
df_trustpilot_negative_common = df_trustpilot_negative[df_trustpilot_negative['Location Name'].isin(common_locations)]
```

```python
print(df_google_negative_common.shape)
print(df_trustpilot_negative_common.shape)
```

```python
# Merge the reviews to form a new list.
merged_reviews = df_google_negative_common['Comment'].tolist() + df_trustpilot_negative_common['Review Content'].tolist()
```

```python
print(len(merged_reviews))
merged_reviews[:3]
```

```python
print(len(df_google_negative_common))
print(len(df_trustpilot_negative_common))
print(len(df_google_negative_common) + len(df_trustpilot_negative_common))
```

```python
# Apply lowercase and remove numbers
def preprocess_for_bertopic(text):
    text = text.lower()
    text = re.sub(r'\d+', '', text)
    return text

merged_reviews_clean = [preprocess_for_bertopic(review) for review in merged_reviews]
```

```python
# Sanity check
print(len(merged_reviews_clean))
merged_reviews_clean[:3]
```

```python
# Remove synonyms from Bert topics
from sklearn.feature_extraction.text import CountVectorizer
vectorizer_model = CountVectorizer(stop_words=list(custom_stopwords), ngram_range=(1, 2))

topic_model = BERTopic(vectorizer_model=vectorizer_model)
topics, probs = topic_model.fit_transform(merged_reviews_clean)
```

```python
# Preprocess this data set. Use BERTopic on this cleaned data set.
from bertopic import BERTopic
from sklearn.feature_extraction.text import CountVectorizer

vectorizer_model = CountVectorizer(stop_words=list(custom_stopwords), ngram_range=(1, 2))
topic_model = BERTopic(vectorizer_model=vectorizer_model)

topics, probs = topic_model.fit_transform(merged_reviews_clean)
```

```python
# List top topics with document frequencies
topic_info = topic_model.get_topic_info()
topic_info.head(10)
```

```python
# For the top 2 topics, list out the top words.
top_topics = topic_info[topic_info['Topic'] != -1].head(2)['Topic'].tolist()

for topic_num in top_topics:
    print(f"Topic {topic_num}:")
    print(topic_model.get_topic(topic_num))
    print()
```

There are some new words like "air con" related to hot temperatures at the gym.

```python
print(topic_info.head(5))
```

```python
outlier_docs = [merged_reviews_clean[i] for i, t in enumerate(topics) if t == -1]
print(len(outlier_docs))
print(outlier_docs[:5])
```

```python
total_docs = topic_info['Count'].sum()
outlier_pct = 1558 / total_docs * 100
print(f"Outlier percentage: {outlier_pct:.1f}%")
```

Topic -1 indicates the outlier for approximately 39.4% of documents in this run. This is expected behaviour in BERTopic, which does not force every document into a topic. The "gym" stopword removal, while successfully removing that word from the vocabulary, does not affect this outlier rate, since document clustering is driven by overall content similarity rather than the presence or absence of any single word.

```python
# See a few real reviews representative of each topic
for topic_num in top_topics:
    print(f"Topic {topic_num} — top words:")
    print(topic_model.get_topic(topic_num))
    print(f"\nRepresentative documents:")
    docs = topic_model.get_representative_docs(topic_num)
    for doc in docs[:3]:
        print(f"- {doc}")
    print("\n" + "="*80 + "\n")
```

```python
def run_llm_chat(messages: list,
                     model: str = "phi4-mini",
                     temperature: float = 0.2) -> str:
    """
    Use ollama.chat;
    """
    response = client.chat(
        model=model,
        messages=messages,
        options={"temperature": temperature, "num_ctx": 2192, "num_predict": 1000},
    )
    return response.message.content
```

```python
# Messages template
messages_template = [
    {
        "role": "system",
        "content": (
            "You are a Data Analyst expert at extracting key topics from customer reviews of a gym business. "
            "Your task is to identify the main issues or subjects discussed in a given review. "
            "Return ONLY a numbered list of up to 3 topics, with each one on a new line. "
            "Each topic MUST be directly derived and grounded in the provided review text. "
            "Do NOT include any explanations, labels, or commentary outside the numbered list."
        )
    },
    {
        "role": "user",
        "content": "Extract up to 3 main topics from this review and return ONLY a numbered list: \"The air conditioning was broken all summer and the changing rooms smelled terrible.\""
    },
    {
        "role": "assistant",
        "content": "1. Broken air conditioning\n2. Poor changing room hygiene\n3. Facility maintenance issues"
    },
]

def build_messages(review_text):
    return messages_template + [
        {
            "role": "user",
            "content": f"Extract up to 3 main topics from this review and return ONLY a numbered list: {review_text}"
        }
    ]
```

```python
# Add the following prompt to every review, before passing it on to the model. Run the model.
import time

# Test on a small subset first to gauge runtime
test_reviews = all_negative_reviews_clean[:10]

llm_topics_raw = []
start = time.time()

for review in test_reviews:
    messages = build_messages(review)
    output = run_llm_chat(messages)
    llm_topics_raw.append(output)

elapsed = time.time() - start
print(f"Time for {len(test_reviews)} reviews: {elapsed:.2f} seconds")
print(f"Estimated time for all {len(all_negative_reviews_clean)} reviews: {elapsed / len(test_reviews) * len(all_negative_reviews_clean) / 60:.2f} minutes")
```

```python
# Append each of these topics from each review to create a comprehensive list.
import re

def parse_numbered_topics(output_text):
    """
    Parses a numbered list like '1. Topic one\n2. Topic two\n3. Topic three'
    into a Python list of topic strings.
    """
    lines = output_text.strip().split('\n')
    topics = []
    for line in lines:
        # Remove leading number, period, and whitespace (e.g. "1. Topic" -> "Topic")
        cleaned = re.sub(r'^\d+\.\s*', '', line).strip()
        if cleaned:
            topics.append(cleaned)
    return topics
```

```python
# The output of the model will be the top 3 topics from each review. Append each of these topics from each review to create a comprehensive list
comprehensive_topic_list = []
failed_reviews = []

sample_size = 500

for i, review in enumerate(all_negative_reviews_clean[:sample_size]):
    try:
        messages = build_messages(review)
        output = run_llm_chat(messages)
        topics = parse_numbered_topics(output)
        comprehensive_topic_list.extend(topics)
    except Exception as e:
        failed_reviews.append((i, str(e)))
        continue

    if i % 100 == 0:
        print(f"Processed {i}/{sample_size} reviews — {len(comprehensive_topic_list)} topics so far")

print(f"Completed: {len(comprehensive_topic_list)} topics extracted")
print(f"Failed: {len(failed_reviews)} reviews")
```
