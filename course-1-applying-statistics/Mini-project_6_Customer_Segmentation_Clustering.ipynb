# **3. Perform feature engineering as follows:**

*   Create new features for frequency, recency, CLV, average unit cost, and customer age.
*   Perform feature scaling and encoding if needed.

```python
# Create new features for frequency, and recency.
ecom_data['Frequency'] = ecom_data.groupby('Customer ID')['Order ID'].transform('count')

ecom_data['Delivery_Date'] = pd.to_datetime(ecom_data['Delivery_Date'])
ecom_data['Recency'] = (pd.to_datetime('today') - ecom_data['Delivery_Date']).dt.days

# CLV
def currency_to_float(currency_str):
    currency_str = str(currency_str).replace('$', '').replace(',', '')
    if '(' in currency_str and ')' in currency_str:
        return -float(currency_str.replace('(', '').replace(')', ''))
    return float(currency_str)

ecom_data['Total Revenue'] = ecom_data['Total Revenue'].apply(currency_to_float)
ecom_data['Profit'] = ecom_data['Profit'].apply(currency_to_float)
ecom_data['CLV'] = ecom_data['Profit'] * ecom_data['Quantity']

# Create the Avg Unit Cost.
ecom_data['Avg_Unit_Cost'] = ecom_data['Total Revenue'] / ecom_data['Quantity']

# Customer age
ecom_data['DOB'] = pd.to_datetime(ecom_data['Customer_BirthDate'], format='%d%b%Y')
today = pd.to_datetime('now')
ecom_data['Age'] = (today - ecom_data['DOB']).dt.days // 365

print(ecom_data.shape)
print(ecom_data.head())
```

To be able to perform the CLV it needs to get the profit on the CLV formula. Therefore the total of extra features is 7 but the extra one won't be used for the rest of the project.

```python
# Customer aggregation by ID
ecom_data_agg = ecom_data.groupby('Customer ID').agg({'Frequency': 'count', 'Recency': 'min','CLV': 'sum', 'Avg_Unit_Cost': 'mean','Age': 'max'}).reset_index()


# Display results
print(ecom_data_agg.shape)
ecom_data_agg.head()
```

```python
# Print Silhouette Score
from sklearn.metrics import silhouette_score, silhouette_samples
import matplotlib.pyplot as plt
import numpy as np

# Define range_n_clusters here as the prvious plot results
range_n_clusters = range(2, 11)

for num_clusters in range_n_clusters:
  fig, (ax1, ax2) = plt.subplots(1, 2)
  fig.set_size_inches(18, 7)

  # The silhouette coefficient
  ax1.set_xlim([-0.1, 1])
  ax1.set_ylim([0, len(ecom_data_agg_transformed) + (num_clusters + 1) * 10])

  # Initialize the clusterer
  clusterer = KMeans(n_clusters=num_clusters, random_state=10, n_init=10)
  cluster_labels = clusterer.fit_predict(ecom_data_agg_transformed)

  # Silhouette average score
  silhouette_avg = silhouette_score(ecom_data_agg_transformed, cluster_labels)
  print("For n_clusters =", num_clusters, "The average silhouette_score is :", silhouette_avg)

  # Compute the silhouette scores for each sample
  sample_silhouette_values = silhouette_samples(ecom_data_agg_transformed, cluster_labels)

  y_lower = 10
  for i in range(num_clusters):
      ith_cluster_silhouette_values = sample_silhouette_values[cluster_labels == i]
      ith_cluster_silhouette_values.sort()
      size_cluster_i = ith_cluster_silhouette_values.shape[0]
      y_upper = y_lower + size_cluster_i
      color = plt.cm.nipy_spectral(float(i) / num_clusters)
      ax1.fill_betweenx(
          np.arange(y_lower, y_upper),
          0,
          ith_cluster_silhouette_values,
          facecolor=color,
          edgecolor=color,
          alpha=0.7,
      )

      # Label the silhouette plots with their cluster numbers at the middle
      ax1.text(-0.05, y_lower + 0.5 * size_cluster_i, str(i))
      y_lower = y_upper + 10

  ax1.set_title("The silhouette plot for the various clusters.")
  ax1.set_xlabel("The silhouette coefficient values")
  ax1.set_ylabel("Cluster label")

  # The vertical line for average silhouette scores
  ax1.axvline(x=silhouette_avg, color="red", linestyle="--")
  ax1.set_yticks([])
  ax1.set_xticks([-0.1, 0, 0.2, 0.4, 0.6, 0.8, 1])

  # 2nd Plot showing the actual clusters formed
  colors = plt.cm.nipy_spectral(cluster_labels.astype(float) / num_clusters)
  ax2.scatter(
      ecom_data_agg_transformed[:, 0], ecom_data_agg_transformed[:, 1], marker=".", s=30, lw=0, alpha=0.7, c=colors, edgecolor="k"
  )

  # Labeling the clusters
  centers = clusterer.cluster_centers_
  ax2.scatter(
      centers[:, 0],
      centers[:, 1],
      marker="o",
      c="white",
      alpha=1,
      s=200,
      edgecolor="k",
  )

  for i, c in enumerate(centers):
      ax2.scatter(c[0], c[1], marker=f"${i}$", alpha=1, s=50, edgecolor="k")

  ax2.set_title("The visualization of the clustered data.")
  ax2.set_xlabel("Feature space for the 1st feature")
  ax2.set_ylabel("Feature space for the 2nd feature")

  plt.suptitle(
      "Silhouette analysis for KMeans clustering on sample data with n_clusters = %d" % num_clusters,
      fontsize=14, fontweight="bold",
  )
  plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import AgglomerativeClustering
from scipy.cluster.hierarchy import dendrogram, linkage

# Function to create a dendrogram using a given linkage metric
def plot_dendrogram(small_ecom, method, ax, truncate_mode=None, p=None):
    linked = linkage(small_ecom, method=method)
    if truncate_mode and p:
        dendrogram(linked, ax=ax, truncate_mode=truncate_mode, p=p, show_leaf_counts=True, leaf_rotation=90., leaf_font_size=8.)
    else:
        dendrogram(linked, ax=ax)
    ax.set_title(f'Dendrogram ({method.capitalize()} Linkage)')

# Create plotting setup - 1 row, 5 columns to compare methods
fig, axes = plt.subplots(1, 5, figsize=(20, 8))

# Plot the original scatter plot
axes[0].scatter(small_ecom[:, 0], small_ecom[:, 1], alpha=0.6)
axes[0].set_title('Original E-commerce Data')
axes[0].set_xlabel('Feature 1')
axes[0].set_ylabel('Feature 2')

# Define the linkage methods to compare
linkage_methods = ['single', 'complete', 'average', 'ward']

# Plot dendrograms for each linkage method in the subsequent columns
for i, method in enumerate(linkage_methods):
    plot_dendrogram(small_ecom, method, axes[i+1], truncate_mode='lastp', p=20)

# Display the comparison plots
plt.tight_layout()
plt.show()
```
