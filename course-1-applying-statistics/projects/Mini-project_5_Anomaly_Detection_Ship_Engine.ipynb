- **Lubrication oil temperature:** High lubrication oil temperature suggests the oil is overheating, which can degrade its lubricating properties and lead to engine damage. A low lubrication oil temperature may indicate it is not reaching its optimal operating temperature, potentially causing inadequate lubrication.
- **Coolant temperature:** High coolant temperature signals overheating, which various issues, including a failed thermostat, coolant leak, or insufficient coolant flow can cause. A low coolant temperature could suggest the engine is not reaching its optimal operating temperature, affecting performance and efficiency.

Issues with engines could lead to engine malfunctions, potential safety hazards, and downtime (e.g. delayed deliveries), resulting in the breakdown of a ship’s overall functionality, consequently impacting the business, such as affecting revenue via failure to deliver goods. # **5. Perform anomaly detection with a statistical method and identify possible anomalies.**

*   Use the interquartile range (IQR) method to identify outliers for each feature.
*   Create a new column (corresponding to each feature) that will indicate (in binary – 0,1) if the value of that feature is an outlier as per IQR calculations.
*   Use IQR to identify the number of features that must simultaneously be in outlier condition, in order for a sample to be classified as an outlier, such that the total percentage of samples identified as outliers falls within the 1-5% range.
*   Record your thoughts and observations.

### IQR method




```python
# IQR function to identify any outliers.
def iqr(column):
    return column.quantile(0.75) - column.quantile(0.25)


# Print the IQR output.
iqr_values = (ship_df[["Engine rpm", "Lub oil pressure", "Fuel pressure", "Coolant pressure", "lub oil temp", "Coolant temp" ]])
print(iqr_values.shape)
ship_df.head()
```

```python
# Calculate the quantiles.
q1 = ship_df.quantile(0.25)
q3 = ship_df.quantile(0.75)

# View the output.
print("The value of Q1 is:", q1)
print("The value of Q3 is:", q3)

# View the IQR.
iqr = q3 - q1
print("The value of IQR is:",iqr)
```

```python
# Filtering the outliers and calculate the upper and lower limits.
lower = q1-1.5*iqr
upper = q3+1.5*iqr

# View the limits.
print("Upper limit: ", upper)
print("Lower limit: ", lower)
```

```python
# Create the binary columns for each feature.
column_stack = ["Engine rpm", "Lub oil pressure", "Fuel pressure", "Coolant pressure", "lub oil temp", "Coolant temp"]

for column in column_stack:
  q1 = ship_df[column].quantile(0.25)
  q3 = ship_df[column].quantile(0.75)
  iqr = q3 - q1

  lower = q1 - 1.5*iqr
  upper = q3 + 1.5*iqr

  ship_df[column + "_outlier"] = ((ship_df[column] < lower) | (ship_df[column] > upper)).astype(int)

# View the new columns.
print(ship_df.shape)
ship_df.head()
```

```python
# Identify which observations have flagged columns (sum up the rows).
outlier_columns = [column for column in ship_df.columns if "_outlier" in column]
outlier_conts = ship_df[outlier_columns].sum()

# View the results.
print(outlier_conts)
```

```python
# Display the IQR results as boxplot with individual y-axis scales.
fig, axes = plt.subplots(1, 6, figsize=(18, 6))

for i, column in enumerate(iqr_values.columns):
    axes[i].boxplot(iqr_values[column].values, patch_artist=True, showmeans=True)
    axes[i].set_title(column, fontsize=9)
    axes[i].set_ylabel('Values')
    axes[i].set_xticks([])  # Remove the '1' tick label
    axes[i].grid(True, alpha=0.3, axis='y')

plt.suptitle('Box Plot - IQR Visualization')
plt.tight_layout()
plt.show()
```

```python
sns.boxplot(data=ship_df, orient='h')
plt.show()
```

```python
# Calculate the percentage of the dataset per feature.
outlier_percentages = (outlier_conts / len(ship_df)) * 100

# View the results.
print(outlier_percentages)
```

```python
# Count outliers rows.
ship_df['outlier_sum'] = ship_df[outlier_columns].sum(axis=1)
print(ship_df['outlier_sum'].value_counts().sort_index())
```

```python
# Calculate the percentage of outliers rows.
total = len(ship_df)
summary = []

for threshold in [1, 2, 3, 4, 5]:
    count = (ship_df['outlier_sum'] >= threshold).sum()
    percentage = (count / total) * 100
    summary.append({'Simultaneous Features': f"{threshold}+",
                    'Count of Outliers': count,
                    'Percentage of Outliers (%)': round(percentage, 2)})

iqr_summary_table = pd.DataFrame(summary)
print(iqr_summary_table)
```

The IQR method identified outliers by row, making it difficult to detect anomalies, as the data has 19k rows. Therefore, this is not the best method for this dataset to identify anomalies. However, we identified the number of observations with outliers and concluded that features such as Fuel Pressure (5.81%) and Lub oil (13.39%) exceed our target limit, potentially indicating they need maintenance to prevent ship engine issues.


When analysed by rows, we can identify that 23.73% had more than 1+ feature with anomalies, which is above the 1-5% initial suggested goal.
The others (2+ to 5+) are below 2.16%, suggesting a few anomalies.

At 2+, 2.16% samples were flagged; at 3+, only 0.06% are flagged as outliers. The 4+ and 5+ have no samples flagged.

However, the majority of the features still have anomalies. The anomalies may indicate a potential engine issue, or they may be normal due to acceleration. Further investigation with the domain expert is needed to properly address those outliers.

```python
# Visualise the data
def plot_embedding(X_scaled, predictions, title):
    # 2D using PCA.
    pca = PCA(n_components=2)
    data_2d = pca.fit_transform(X_scaled)

    # Plot.
    plt.figure(figsize=(10,10))
    plt.title(f"{title}\n(PCA: {pca.explained_variance_ratio_[0]:.1%} + {pca.explained_variance_ratio_[1]:.1%} = {sum(pca.explained_variance_ratio_):.1%} variance explained)")

    # Plot data points.
    sns.scatterplot(x=data_2d[:, 0], y=data_2d[:, 1],
                    hue=np.where(predictions == 1, 'normal', 'outlier'),
                    style=np.where(predictions == -1, 'outlier', 'normal'),
                    palette={'normal': 'blue', 'outlier': 'red'},
                    alpha=0.6, edgecolor='k')

    # Plot settings.
    plt.xlabel(f'First Principal Component ({pca.explained_variance_ratio_[0]:.1%} variance)')
    plt.ylabel(f'Second Principal Component ({pca.explained_variance_ratio_[1]:.1%} variance)')
    plt.show()
```

Once gamma is reduced to 0.5 again and Nu is also reduced, include the data points as normal. What we are trying to achieve here is find a balance across the tests among the Nu and Gamma options and determine which option is best.

```python
# Decrease the Nu value to 0.1.
plot_embedding_011 = OneClassSVM(kernel='rbf', gamma=0.5, nu=0.01)
plot_embedding_011.fit(X_scaled)

# Get predictions first.
predictions_011 = plot_embedding_011.predict(X_scaled)

# Plot the gamma 1.
plot_embedding(X_scaled, predictions_011, 'One-Class SVM With Anomaly Detection - 0.1 Nu')
```

```python
# Calculate the percentage of the outliers for the model.
# Test different models.
gamma_values = [0.1, 0.5, 1.0]
nu_values = [0.01, 0.05, 0.1]

results = []

for gamma in gamma_values:
    for nu in nu_values:
        model = OneClassSVM(gamma=gamma, nu=nu)
        model.fit(X_scaled)
        predictions = model.predict(X_scaled)

        n_outliers = (predictions == -1).sum()
        outlier_pct = (n_outliers / len(predictions)) * 100

        results.append({
            'Gamma': gamma,
            'Nu': nu,
            'Count Outliers': n_outliers,
            'Percentage of Outlliers': f"{outlier_pct:.1f}%"
        })

# Display as table.
df = pd.DataFrame(results)
print(df)
```

Among all OneClass SVM experiments, rows 0 (Gamma=0.1, Nu=0.01) and 1 (Gamma=0.1, Nu=0.05) share the same low Gamma value but differ in Nu, yielding 1% and 5% outlier rates, respectively, both within the acceptable range for this project. However, both configurations still classify some potential outliers as normal data points, which may be acceptable for the engine RPM feature but could be problematic for other features.

Row 1 (Gamma=0.1, Nu=0.05) was selected because its Nu=0.05 is the maximum within the acceptable range for this project, which improves anomaly detection and makes the approach more robust. In contrast, row 0's very low Nu of 0.01 is overly conservative, flagging only 189 outliers compared with the 978 identified by row 1, increasing the risk of missing genuine anomalies by treating them as normal data points. This has practical implications beyond the model itself, as failing to detect real anomalies could influence operational decisions and affect other stakeholders. Domain expert feedback would nonetheless be beneficial to validate whether the flagged points represent true anomalies across all features.
